```python
#selecting interesting events: I picked the largest flare, the smallest flare and one
#without a goes class (also an M class flare for comparison):
largeflare = flares.loc[21278]
medflare = flares.loc[13953]
nogoesclass = flares.loc[12921]
smallflare = flares.loc[14252]

print(largeflare["peak_UTC"], medflare["peak_UTC"], nogoesclass["peak_UTC"], smallflare["peak_UTC"])
```

```python
#using STIX and GOES to get lightcurves for the events and compare:
from sunpy.net import Fido, attrs as a
from stixpy.net.client import STIXClient

import sunpy.timeseries as ts
from stixpy.timeseries import quicklook 

import matplotlib.pyplot as plt
from datetime import datetime
```

```python
#large flare:
lf_query = Fido.search(a.Time(largeflare["start_UTC"], largeflare["end_UTC"]), 
                       a.Instrument.stix, 
                       a.stix.DataProduct.ql_lightcurve)

lf_query
```

```python
lf_file = Fido.fetch(lf_query)
```

```python
lf_lc = ts.TimeSeries(lf_file)
```

```python
lf_lc.plot()
plt.xlim(datetime(2024, 10, 3, 11, 50), datetime(2024, 10, 3, 12, 50))
plt.axvline(largeflare["peak_UTC"], 0, 1, color = 'grey', linestyle = 'dashed')
plt.savefig("C:/Users/derva/OneDrive/Documents/DIAS/solar-orbiter-flare-statistics/figures/stix_flarelist_largeflare.png")
```

```python
goes_lf_result = Fido.search(a.Time(largeflare["start_UTC"], largeflare["end_UTC"]),
                     a.Instrument("XRS"),
                     a.goes.SatelliteNumber(16))

goes_lf_files = Fido.fetch(goes_lf_result)

goes_lf_result
```

```python
goes_lf = ts.TimeSeries(goes_lf_files, concatenate=True)
```

```python
goes_lf.plot()
plt.xlim(datetime(2024, 10, 3, 11, 50), datetime(2024, 10, 3, 12, 50))
plt.axvline(datetime(2024, 10, 3, 12, 12, 37), 0, 1, color = 'grey', linestyle = 'dashed', label = 'Peak from STIX flare list')
plt.legend()
plt.savefig("C:/Users/derva/OneDrive/Documents/DIAS/solar-orbiter-flare-statistics/figures/stix_flarelist_largeflare_goes.png")
```
